```python
df.columns=['recipe_id', 'food_name', 'ingred_view', 'ingred']
```

```python
df['ingred_view']=df['ingred_view'].str.replace("", " ", regex=True)
```

```python
import re

# 모든 문자열을 하나로 합치기
all_text = ' '.join(df['ingred_view'].astype(str))

# 특수문자만 추출
special_chars = re.findall(r'[^a-zA-Z0-9가-힣\s]', all_text)
# 중복 제거 및 정렬
unique_special_chars = sorted(set(special_chars))

print(unique_special_chars)
```
